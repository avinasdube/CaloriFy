# Calorie Regression Notebook

This notebook explores the regression techniques used to predict calorie content based on food images. We will load the trained regression model, preprocess the test images, and evaluate the model's performance.

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import mean_squared_error, r2_score
from src.calorie_regressor import load_model, preprocess_image

# Load calorie mapping data
calorie_mapping = pd.read_csv('../data/calorie_mapping.csv')
calorie_mapping.head()
```

```python
# Load the trained calorie regressor model
model = load_model('../models/calorie_regressor.pkl')

# Function to predict calories for a given image
def predict_calories(image_path):
    image = preprocess_image(image_path)
    prediction = model.predict(image)
    return prediction[0]

# Test the prediction function with a sample image
sample_image_path = '../data/test_images/sample_food.jpg'
predicted_calories = predict_calories(sample_image_path)
print(f'Predicted calories: {predicted_calories:.2f} kcal')
```

```python
# Evaluate model performance on test dataset
test_images = [...]  # List of test image paths
true_calories = [...]  # Corresponding true calorie values

predictions = [predict_calories(img) for img in test_images]

# Calculate performance metrics
mse = mean_squared_error(true_calories, predictions)
r2 = r2_score(true_calories, predictions)

print(f'Mean Squared Error: {mse:.2f}')
print(f'R^2 Score: {r2:.2f}')
```

```python
# Visualize predictions vs true values
plt.figure(figsize=(10, 6))
sns.scatterplot(x=true_calories, y=predictions)
plt.xlabel('True Calories')
plt.ylabel('Predicted Calories')
plt.title('True vs Predicted Calories')
plt.plot([0, max(true_calories)], [0, max(true_calories)], 'r--')  # Diagonal line
plt.show()
```

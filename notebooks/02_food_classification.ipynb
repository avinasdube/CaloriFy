# Food Classification

In this notebook, we will explore food classification techniques using a Convolutional Neural Network (CNN). We will load the dataset, preprocess the images, train the model, and evaluate its performance.

```python
# Import necessary libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from keras.preprocessing.image import ImageDataGenerator
from keras.models import load_model
from src.food_classifier import FoodClassifier
from src.image_preprocessing import preprocess_images

# Set random seed for reproducibility
np.random.seed(42)
```

```python
# Load and preprocess the dataset
data = pd.read_csv('../data/calorie_mapping.csv')
images, labels = preprocess_images('../data/raw')

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(images, labels, test_size=0.2, random_state=42)
```

```python
# Data augmentation
datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Fit the model
model = FoodClassifier()
model.train(X_train, y_train, datagen)
```

```python
# Evaluate the model
test_loss, test_accuracy = model.evaluate(X_test, y_test)
print(f'Test accuracy: {test_accuracy:.2f}')
```

```python
# Visualize some predictions
predictions = model.predict(X_test)
plt.figure(figsize=(12, 12))
for i in range(9):
    plt.subplot(3, 3, i + 1)
    plt.imshow(X_test[i])
    plt.title(f'Predicted: {predictions[i]}')
    plt.axis('off')
plt.show()
```
